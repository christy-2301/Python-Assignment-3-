***Number Guessing Game***

Here we are creating a simple number guessing game

* All user has 3 attempts ,**attempts=3**.
* Each time the user plays the no. of tries dereases **attempts -= 1**.
* If the number entered by the user is less than 1 or greater than 10 **if guess < 1 or guess > 10:**.
*Prints **"Your guess is out of range. Please guess a number between 1 and 10."**.
* Loop continues.
* And if the the number guessed by the user is correct **guess ==  secret_number**.
* Print **congratulations! you guessed the correct number**.
* Loop breaks if the number was guessed correctly.
* If the number guessed is too low **guess < secret_number**.
* If the number guessed is too  high **guess > secret_number**.
* Even after all the tries if the used cannot guess the secret number prints **better luck next time**.




```python
import random
secret_number = random.randint(1, 10)
attempts = 3

while attempts > 0:
  guess = int(input("Guess the number (between 1 and 10): "))
  attempts -= 1

  if guess < 1 or guess > 10:
   print("Your guess is out of range. Please guess a number between 1 and 10.")
   continue

  if guess == secret_number:
   print("Congratulations! You guessed the correct number.")
   break

  if guess > secret_number:
   print("Too high, Try again")

  if guess < secret_number:
   print("Too low , Try again")

else:
 print("Better luck next time!")
```

***Multiplication Table Generator***

Here we are creating a multiplication table generator


*   We are taking input from the user and creating that significant numbers multiplication table
* In for loop the range is set **range(1,11)** to get 1 to 10.
* And the table is printed in the format the ** number x i = input number * i**



```python
number = int(input("Enter the number for which you want the multiplication table:"))
for i in range(1,11):
  print(number, "x",i ,"=", number * i)

```

***BMI Calculator***

Here we are creating a simple BMI calculator using a function

* We create a function **calculator_bmi(weight,height)**.
* User enters the weight in **kg** and height in **meter.**and they are stored in  variables   
* To find bmi  (bmi= (weight / height**2)).
* The calculated bmi is stored in **BMI**.
*And we use return function to get the calculated value.
*And the calculated bmi is printed.



```python
def calculator_bmi(weight,height):
  BMI= weight / height**2
  return BMI

weight = float(input("Enter your weight in kg:"))
height = float(input("Enter your height in meters:"))
BMI=calculator_bmi(weight,height)
print("Your BMI is:our",BMI)
```
